```python

# Start with the Premier League table
master_df = league_table.copy()

# Datasets to merge
merge_sources = {
    "std": standard,
    "shoot": shooting,
    "gk": goalkeeping,
    "time": playing_time,
    "misc": miscellaneous
}

# Merge each dataset using Squad as the key
for prefix, df in merge_sources.items():

    # Prefix every column except the team name
    renamed_df = df.rename(
        columns={
            col: f"{prefix}_{col}"
            for col in df.columns
            if col != "Squad"
        }
    )

    master_df = master_df.merge(
        renamed_df,
        on="Squad",
        how="left",
        validate="one_to_one"
    )

# Sort by current league position
master_df = master_df.sort_values("Rk").reset_index(drop=True)

# Inspect the merged dataset
print("Master dataset shape:", master_df.shape)
print("Number of teams:", master_df["Squad"].nunique())
print("Duplicate teams:", master_df["Squad"].duplicated().sum())
print("Total missing values:", master_df.isna().sum().sum())

display(master_df.head())
```

```python

# Create a separate feature-engineering dataframe
team_features = master_df.copy()

# Per-match performance
per_match_metrics = {
    "std_Gls": "Attack_Goals_Per_Match",
    "std_Ast": "Attack_Assists_Per_Match",
    "shoot_Sh": "Attack_Shots_Per_Match",
    "shoot_SoT": "Attack_Shots_On_Target_Per_Match",
    "gk_GA": "Defence_Goals_Conceded_Per_Match",
    "gk_Saves": "Goalkeeping_Saves_Per_Match",
    "misc_Int": "Defence_Interceptions_Per_Match",
    "misc_TklW": "Defence_Tackles_Won_Per_Match",
    "misc_CrdY": "Discipline_Yellow_Cards_Per_Match",
    "misc_CrdR": "Discipline_Red_Cards_Per_Match",
    "time_Subs": "Squad_Substitutions_Per_Match"
}

for source, new_name in per_match_metrics.items():
    team_features[new_name] = (
        team_features[source] / team_features["MP"]
    )

# Display the newly engineered features
display(
    team_features[
        ["Squad"] + list(per_match_metrics.values())
    ].round(2)
)
```

```python

# Metrics where higher values indicate better performance
higher_is_better = [
    "Attack_Goals_Per_Match",
    "Attack_Assists_Per_Match",
    "Attack_Shots_On_Target_Per_Match",
    "shoot_G/Sh",
    "gk_Save%",
    "gk_CS%"
]

# Metrics where lower values indicate better performance
lower_is_better = [
    "Defence_Goals_Conceded_Per_Match",
    "gk_GA90"
]

# Create normalized percentile scores
for col in higher_is_better:
    team_features[col + "_Score"] = (
        team_features[col].rank(pct=True) * 100
    )

for col in lower_is_better:
    team_features[col + "_Score"] = (
        (1 - team_features[col].rank(pct=True)) * 100
    )

# Display normalized scores
score_columns = [
    col for col in team_features.columns
    if col.endswith("_Score")
]

display(
    team_features[
        ["Squad"] + score_columns
    ].sort_values(
        "Attack_Goals_Per_Match_Score",
        ascending=False
    ).round(1)
)
```

```python

# 1. Attack Rating
team_features["Attack_Rating"] = (
    team_features["Attack_Goals_Per_Match_Score"] * 0.35
    + team_features["Attack_Assists_Per_Match_Score"] * 0.20
    + team_features["Attack_Shots_On_Target_Per_Match_Score"] * 0.25
    + team_features["shoot_G/Sh_Score"] * 0.20
)

# 2. Defence Rating
team_features["Defence_Rating"] = (
    team_features["Defence_Goals_Conceded_Per_Match_Score"] * 0.60
    + team_features["gk_CS%_Score"] * 0.40
)

# 3. Goalkeeping Rating
team_features["Goalkeeping_Rating"] = (
    team_features["gk_Save%_Score"]
)

# Display ratings
ratings_preview = team_features[
    ["Squad", "Attack_Rating", "Defence_Rating", "Goalkeeping_Rating"]
].sort_values("Attack_Rating", ascending=False)

display(ratings_preview.round(2))
```

```python

# Convert the last-five results into points
result_points = {
    "W": 3,
    "D": 1,
    "L": 0
}

def calculate_last5_points(form):
    results = str(form).replace(" ", "").split()
    results = list(str(form).replace(" ", "").replace("\n", ""))

    return sum(result_points.get(result, 0) for result in results[-5:])

team_features["Last5_Points"] = (
    team_features["Last 5"].apply(calculate_last5_points)
)

# Convert points into a relative percentile rating
team_features["Current_Form_Rating"] = (
    team_features["Last5_Points"].rank(pct=True) * 100
)

# Display current form
form_preview = team_features[
    ["Squad", "Last 5", "Last5_Points", "Current_Form_Rating"]
].sort_values(
    "Current_Form_Rating",
    ascending=False
).reset_index(drop=True)

display(form_preview.round(2))
```

```python

# Map FPL club names to FBref club names
club_mapping = {
    "Leeds": "Leeds United",
    "Man City": "Manchester City",
    "Man Utd": "Manchester Utd",
    "Nott'm Forest": "Nottingham",
    "Spurs": "Tottenham"
}

# Standardise FPL club names
fpl_merged["team"] = fpl_merged["team"].replace(club_mapping)

# Validate club matching
fpl_clubs = set(fpl_merged["team"].dropna().unique())
fbref_clubs = set(master_df["Squad"].dropna().unique())

print("FPL clubs:", len(fpl_clubs))
print("FBref clubs:", len(fbref_clubs))

print("\nRemaining unmatched FPL clubs:")
print(sorted(fpl_clubs - fbref_clubs))

print("\nRemaining unmatched FBref clubs:")
print(sorted(fbref_clubs - fpl_clubs))
```

```python

# Sort players by cumulative FPL points within each club
fpl_sorted = fpl_merged.sort_values(
    ["team", "total_points"],
    ascending=[True, False]
).copy()

# Calculate team-level FPL indicators
fpl_team_summary = fpl_sorted.groupby("team").agg(
    FPL_Squad_Size=("id", "nunique"),
    FPL_Total_Points=("total_points", "sum"),
    FPL_Average_Price=("now_cost_player", "mean")
)

# Top 11 players
top11 = (
    fpl_sorted.groupby("team")
    .head(11)
    .groupby("team")["total_points"]
    .agg(["sum", "mean"])
    .rename(columns={
        "sum": "FPL_Top11_Points",
        "mean": "FPL_Top11_Average"
    })
)

# Top 5 players
top5 = (
    fpl_sorted.groupby("team")
    .head(5)
    .groupby("team")["total_points"]
    .sum()
    .rename("FPL_Top5_Points")
)

# Combine indicators
fpl_team_summary = (
    fpl_team_summary
    .join(top11)
    .join(top5)
    .reset_index()
    .rename(columns={"team": "Squad"})
)

display(
    fpl_team_summary.sort_values(
        "FPL_Top11_Points",
        ascending=False
    ).round(2)
)

print("\nSummary shape:", fpl_team_summary.shape)
```

```python

# Merge FPL team indicators into the team feature dataset
team_features = team_features.merge(
    fpl_team_summary,
    on="Squad",
    how="left",
    validate="one_to_one"
)

print("Updated shape:", team_features.shape)
print("Teams:", team_features["Squad"].nunique())
print("\nMissing FPL values:")
print(
    team_features[
        ["FPL_Squad_Size", "FPL_Total_Points",
         "FPL_Top11_Points", "FPL_Top11_Average"]
    ].isna().sum()
)

display(
    team_features[
        ["Squad", "Attack_Rating", "Defence_Rating",
         "Goalkeeping_Rating", "Current_Form_Rating",
         "FPL_Top11_Points", "FPL_Top11_Average"]
    ].sort_values("FPL_Top11_Points", ascending=False)
)
```

```python

# STEP 52: RELIABILITY-ADJUSTED PLAYER STATISTICS

# Collect every unique metric used in our rating system
all_metrics = list(dict.fromkeys(
    metric
    for categories in rating_metrics.values()
    for metrics in categories.values()
    for metric in metrics
))

# Create adjusted versions of the metrics
for metric in all_metrics:

    positional_average = (
        outfield_df
        .groupby("position")[metric]
        .transform("mean")
    )

    outfield_df[f"{metric}_adj"] = (
        outfield_df["Reliability"] * outfield_df[metric]
        +
        (1 - outfield_df["Reliability"]) * positional_average
    )

print("Original metrics:", len(all_metrics))

print(
    "Adjusted metrics:",
    sum(f"{metric}_adj" in outfield_df.columns
        for metric in all_metrics)
)

print("\nSAMPLE COMPARISON")

comparison_metrics = [
    "goals_per90",
    "expectedGoals_per90",
    "assists_per90"
]

sample_columns = [
    "player_name",
    "position",
    "minutesPlayed",
    "Reliability"
]

for metric in comparison_metrics:
    if metric in all_metrics:
        sample_columns.extend([
            metric,
            f"{metric}_adj"
        ])

display(
    outfield_df[sample_columns]
    .sort_values("minutesPlayed", ascending=False)
    .head(15)
    .round(3)
)
```

```python

# STEP 56: CONFIDENCE-ADJUSTED ATTRIBUTE SCORES

for position, categories in rating_metrics.items():

    position_mask = outfield_df["position"] == position

    for category, metrics in categories.items():

        percentile_columns = []

        for metric in metrics:

            percentile_col = f"{metric}_raw_percentile"

            outfield_df.loc[position_mask, percentile_col] = (
                outfield_df.loc[position_mask, metric]
                .rank(pct=True) * 100
            )

            percentile_columns.append(percentile_col)

        raw_score = (
            outfield_df.loc[position_mask, percentile_columns]
            .mean(axis=1)
        )

        outfield_df.loc[
            position_mask, f"{category}_Raw_Score"
        ] = raw_score

        outfield_df.loc[
            position_mask, f"{category}_Final_Score"
        ] = (
            50
            + outfield_df.loc[position_mask, "Reliability"]
            * (raw_score - 50)
        )

print("CONFIDENCE-ADJUSTED SCORES")

for position in ["F", "M", "D"]:

    print(f"\nPOSITION: {position}")

    display(
        outfield_df[
            outfield_df["position"] == position
        ][
            [
                "player_name",
                "minutesPlayed",
                "Reliability",
                "Attacking_Raw_Score",
                "Attacking_Final_Score",
                "Creativity_Final_Score",
                "Defending_Final_Score"
            ]
        ]
        .sort_values("Attacking_Final_Score", ascending=False)
        .head(10)
        .round(2)
    )
```

```python
# Create a working copy for the rating engine
rating_df = player_df.copy()

# Convert minutes to numeric
rating_df["minutesPlayed"] = pd.to_numeric(
    rating_df["minutesPlayed"], errors="coerce"
)

# Metrics that should be normalized per 90 minutes
per90_metrics = [
    "goals",
    "expectedGoals",
    "assists",
    "expectedAssists",
    "shotsOnTarget",
    "totalShots",
    "bigChancesCreated",
    "keyPasses",
    "tacklesWon",
    "interceptions",
    "clearances",
    "blockedShots",
    "aerialDuelsWon",
    "ballRecovery"
]

for metric in per90_metrics:
    rating_df[metric] = pd.to_numeric(
        rating_df[metric], errors="coerce"
    )

    rating_df[f"{metric}_per90"] = (
        rating_df[metric] / rating_df["minutesPlayed"] * 90
    )

print("Per-90 metrics created:", len(per90_metrics))
print("New dataset shape:", rating_df.shape)

display(
    rating_df[
        ["player_name", "position", "minutesPlayed",
         "goals_per90", "expectedGoals_per90",
         "assists_per90", "expectedAssists_per90"]
    ].head(10)
)
```

```python
# Attribute metrics
attribute_metrics = {
    "Attack": [
        "goals_per90",
        "expectedGoals_per90"
    ],
    "Creativity": [
        "assists_per90",
        "expectedAssists_per90",
        "keyPasses_per90",
        "bigChancesCreated_per90"
    ],
    "Defending": [
        "tacklesWon_per90",
        "interceptions_per90",
        "clearances_per90",
        "blockedShots_per90",
        "aerialDuelsWon_per90",
        "ballRecovery_per90"
    ]
}

# Verify that all required metrics exist
required_metrics = [
    metric
    for metrics in attribute_metrics.values()
    for metric in metrics
]

missing_metrics = [
    metric for metric in required_metrics
    if metric not in rating_df.columns
]

if missing_metrics:
    raise ValueError(f"Missing metrics: {missing_metrics}")

# Calculate percentile scores within each position
outfield_df = rating_df[
    rating_df["position"].isin(["F", "M", "D"])
].copy()

for attribute, metrics in attribute_metrics.items():

    percentile_columns = []

    for metric in metrics:
        score_column = f"{metric}_percentile"

        outfield_df[score_column] = (
            outfield_df.groupby("position")[metric]
            .rank(pct=True) * 100
        )

        percentile_columns.append(score_column)

    outfield_df[f"{attribute}_Raw"] = (
        outfield_df[percentile_columns].mean(axis=1)
    )

print("Outfield attribute scores calculated.")
display(
    outfield_df[
        [
            "player_name",
            "position",
            "Attack_Raw",
            "Creativity_Raw",
            "Defending_Raw"
        ]
    ].head(10).round(2)
)
```

```python
# Playing-time reliability adjustment
PRIOR_MINUTES = 900

outfield_df["Reliability"] = (
    outfield_df["minutesPlayed"]
    / (outfield_df["minutesPlayed"] + PRIOR_MINUTES)
)

for attribute in ["Attack", "Creativity", "Defending"]:

    outfield_df[f"{attribute}_Rating"] = (
        50
        + outfield_df["Reliability"]
        * (outfield_df[f"{attribute}_Raw"] - 50)
    )

print("RELIABILITY-ADJUSTED ATTRIBUTES")

display(
    outfield_df[
        [
            "player_name",
            "position",
            "minutesPlayed",
            "Reliability",
            "Attack_Raw",
            "Attack_Rating",
            "Creativity_Raw",
            "Creativity_Rating",
            "Defending_Raw",
            "Defending_Rating"
        ]
    ].head(15).round(2)
)
```

```python
# Position-specific attribute weights
position_weights = {
    "F": {
        "Attack_Rating": 0.65,
        "Creativity_Rating": 0.25,
        "Defending_Rating": 0.10
    },
    "M": {
        "Attack_Rating": 0.30,
        "Creativity_Rating": 0.40,
        "Defending_Rating": 0.30
    },
    "D": {
        "Attack_Rating": 0.10,
        "Creativity_Rating": 0.15,
        "Defending_Rating": 0.75
    }
}

def calculate_overall_rating(row):
    weights = position_weights[row["position"]]

    return sum(
        row[attribute] * weight
        for attribute, weight in weights.items()
    )

outfield_df["Overall_Rating"] = outfield_df.apply(
    calculate_overall_rating,
    axis=1
)

outfield_df["Overall_Rating"] = (
    outfield_df["Overall_Rating"].round(2)
)

print("OUTFIELD PLAYER RATINGS CREATED")
print("Total rated players:", len(outfield_df))

for position, label in [
    ("F", "FORWARDS"),
    ("M", "MIDFIELDERS"),
    ("D", "DEFENDERS")
]:
    print(f"\nTOP 10 {label}")

    display(
        outfield_df[outfield_df["position"] == position]
        .sort_values("Overall_Rating", ascending=False)
        [["player_name", "team_name", "minutesPlayed",
          "Attack_Rating", "Creativity_Rating",
          "Defending_Rating", "Overall_Rating"]]
        .head(10)
        .round(2)
    )
```

```python
gk_df = rating_df[rating_df["position"] == "G"].copy()

gk_metrics = [
    "minutesPlayed",
    "goalsPrevented",
    "cleanSheet",
    "saves",
    "savedShotsFromInsideTheBox",
    "savedShotsFromOutsideTheBox",
    "errorLeadToGoal",
    "errorLeadToShot",
    "penaltySave"
]

for col in gk_metrics:
    if col in gk_df.columns:
        gk_df[col] = pd.to_numeric(gk_df[col], errors="coerce")
        print(
            f"{col}: missing={gk_df[col].isna().sum()}, "
            f"available={gk_df[col].notna().sum()}"
        )
    else:
        print(f"{col}: NOT FOUND")

print("\nGoalkeeper sample:")
display(gk_df[gk_metrics].head(10))
```

```python
# Handle the one missing goalsPrevented value
gk_df["goalsPrevented"] = gk_df["goalsPrevented"].fillna(
    gk_df["goalsPrevented"].median()
)

# Calculate goalkeeper metrics per 90
gk_per90_metrics = [
    "goalsPrevented",
    "cleanSheet",
    "errorLeadToGoal",
    "errorLeadToShot"
]

for metric in gk_per90_metrics:
    gk_df[f"{metric}_per90"] = (
        gk_df[metric] / gk_df["minutesPlayed"] * 90
    )

# Calculate position-relative percentile scores
gk_df["ShotStopping_Raw"] = (
    gk_df["goalsPrevented_per90"].rank(pct=True) * 100
)

gk_df["Consistency_Raw"] = (
    gk_df["cleanSheet_per90"].rank(pct=True) * 100
)

# Lower error rates are better, so reverse the ranking
gk_df["ErrorControl_Raw"] = (
    100 - gk_df[
        ["errorLeadToGoal_per90", "errorLeadToShot_per90"]
    ].rank(pct=True).mean(axis=1) * 100
)

# Reliability adjustment
gk_df["Reliability"] = (
    gk_df["minutesPlayed"]
    / (gk_df["minutesPlayed"] + 900)
)

# Shrink attributes towards 50
for attribute in [
    "ShotStopping",
    "Consistency",
    "ErrorControl"
]:
    gk_df[f"{attribute}_Rating"] = (
        50 + gk_df["Reliability"]
        * (gk_df[f"{attribute}_Raw"] - 50)
    )

print("GOALKEEPER ATTRIBUTES CREATED")

display(
    gk_df[
        [
            "player_name",
            "team_name",
            "minutesPlayed",
            "Reliability",
            "ShotStopping_Rating",
            "Consistency_Rating",
            "ErrorControl_Rating"
        ]
    ].head(15).round(2)
)
```

```python
gk_weights = {
    "ShotStopping_Rating": 0.50,
    "Consistency_Rating": 0.30,
    "ErrorControl_Rating": 0.20
}

gk_df["Overall_Rating"] = sum(
    gk_df[column] * weight
    for column, weight in gk_weights.items()
)

gk_df["Overall_Rating"] = gk_df["Overall_Rating"].round(2)

print("TOP 15 GOALKEEPERS")

display(
    gk_df.sort_values(
        "Overall_Rating",
        ascending=False
    )[
        [
            "player_name",
            "team_name",
            "minutesPlayed",
            "ShotStopping_Rating",
            "Consistency_Rating",
            "ErrorControl_Rating",
            "Overall_Rating"
        ]
    ].head(15).round(2)
)
```

```python
# Provisional tactical role scores
role_df = player_ratings.copy()

role_df["Attack_Role_Score"] = (
    0.55 * role_df["Attack_Rating"].fillna(50)
    + 0.35 * role_df["Creativity_Rating"].fillna(50)
    + 0.10 * role_df["Defending_Rating"].fillna(50)
)

role_df["Midfield_Role_Score"] = (
    0.20 * role_df["Attack_Rating"].fillna(50)
    + 0.40 * role_df["Creativity_Rating"].fillna(50)
    + 0.40 * role_df["Defending_Rating"].fillna(50)
)

city = role_df[
    role_df["team_name"] == "Manchester City"
].copy()

# Select three attacking options from forwards and midfielders
attack_candidates = city[
    city["position"].isin(["F", "M"])
]

selected_attack = attack_candidates.nlargest(
    3, "Attack_Role_Score"
)

# Select three remaining midfielders
remaining_midfielders = city[
    (city["position"] == "M")
    & (~city["player_name"].isin(selected_attack["player_name"]))
]

selected_midfield = remaining_midfielders.nlargest(
    3, "Midfield_Role_Score"
)

# Select four defenders
selected_defence = city[
    city["position"] == "D"
].nlargest(4, "Overall_Rating")

# Select one goalkeeper
selected_gk = city[
    city["position"] == "G"
].nlargest(1, "Overall_Rating")

# Combine the provisional XI
city_xi = pd.concat([
    selected_gk,
    selected_defence,
    selected_midfield,
    selected_attack
])

print("MANCHESTER CITY — PROVISIONAL 4-3-3")
display(
    city_xi[
        [
            "player_name",
            "position",
            "minutesPlayed",
            "Overall_Rating",
            "Attack_Role_Score",
            "Midfield_Role_Score"
        ]
    ].round(2)
)
```

```python
team_ratings_list = []

for team in role_df["team_name"].unique():

    squad = role_df[role_df["team_name"] == team]

    # Attack: top 5 forwards and midfielders
    attack_score = (
        squad[squad["position"].isin(["F", "M"])]
        ["Attack_Role_Score"]
        .nlargest(5)
        .mean()
    )

    # Midfield: top 5 midfielders
    midfield_score = (
        squad[squad["position"] == "M"]
        ["Midfield_Role_Score"]
        .nlargest(5)
        .mean()
    )

    # Defence: top 5 defenders
    defence_score = (
        squad[squad["position"] == "D"]
        ["Overall_Rating"]
        .nlargest(5)
        .mean()
    )

    # Goalkeeping: highest-rated goalkeeper
    goalkeeper_score = (
        squad[squad["position"] == "G"]
        ["Overall_Rating"]
        .max()
    )

    team_ratings_list.append({
        "team_name": team,
        "Attack_Rating": attack_score,
        "Midfield_Rating": midfield_score,
        "Defence_Rating": defence_score,
        "Goalkeeping_Rating": goalkeeper_score
    })

team_ratings = pd.DataFrame(team_ratings_list)

print("TEAM STRENGTH — INITIAL RESULTS")

display(
    team_ratings.sort_values(
        "Attack_Rating",
        ascending=False
    ).round(2)
)
```

```python
depth_results = []

for team in role_df["team_name"].unique():

    squad = role_df[role_df["team_name"] == team]

    units = {
        "Attack": (
            squad[squad["position"].isin(["F", "M"])]
            ["Attack_Role_Score"], 5
        ),
        "Midfield": (
            squad[squad["position"] == "M"]
            ["Midfield_Role_Score"], 5
        ),
        "Defence": (
            squad[squad["position"] == "D"]
            ["Overall_Rating"], 5
        ),
        "Goalkeeping": (
            squad[squad["position"] == "G"]
            ["Overall_Rating"], 2
        )
    }

    result = {"team_name": team}

    for unit, (scores, top_n) in units.items():

        sorted_scores = scores.sort_values(ascending=False)

        core = sorted_scores.head(top_n).mean()
        depth = sorted_scores.iloc[top_n:top_n + top_n].mean()

        # If insufficient depth data, use the available core score
        if pd.isna(depth):
            depth = core

        result[f"{unit}_Core"] = core
        result[f"{unit}_Depth"] = depth
        result[f"{unit}_Final"] = 0.70 * core + 0.30 * depth

    depth_results.append(result)

team_ratings = pd.DataFrame(depth_results)

display(
    team_ratings.sort_values(
        "Attack_Final",
        ascending=False
    ).round(2)
)
```

```python
# Check player overlap between attack and midfield selections
overlap_results = []

for team in role_df["team_name"].unique():

    squad = role_df[role_df["team_name"] == team]

    attack_players = set(
        squad[squad["position"].isin(["F", "M"])]
        .nlargest(5, "Attack_Role_Score")["player_name"]
    )

    midfield_players = set(
        squad[squad["position"] == "M"]
        .nlargest(5, "Midfield_Role_Score")["player_name"]
    )

    goalkeeper_count = len(
        squad[squad["position"] == "G"]
    )

    overlap_results.append({
        "team_name": team,
        "Attack_Midfield_Overlap": len(
            attack_players.intersection(midfield_players)
        ),
        "Goalkeepers_Available": goalkeeper_count
    })

overlap_audit = pd.DataFrame(overlap_results)

display(
    overlap_audit.sort_values(
        "Attack_Midfield_Overlap",
        ascending=False
    )
)
```

```python
team_xi_results = []

for team in role_df["team_name"].unique():

    squad = role_df[role_df["team_name"] == team].copy()

    # 1. Select one goalkeeper
    selected_gk = squad[
        squad["position"] == "G"
    ].nlargest(1, "Overall_Rating")

    # 2. Select four defenders
    selected_defence = squad[
        squad["position"] == "D"
    ].nlargest(4, "Overall_Rating")

    # 3. Select three attacking players
    attack_candidates = squad[
        squad["position"].isin(["F", "M"])
    ]

    selected_attack = attack_candidates.nlargest(
        3, "Attack_Role_Score"
    )

    # 4. Select three remaining midfielders
    remaining_midfielders = squad[
        (squad["position"] == "M")
        & (~squad["player_name"].isin(
            selected_attack["player_name"]
        ))
    ]

    selected_midfield = remaining_midfielders.nlargest(
        3, "Midfield_Role_Score"
    )

    # Combine the unique-player selection
    selected_players = pd.concat([
        selected_gk,
        selected_defence,
        selected_midfield,
        selected_attack
    ])

    team_xi_results.append(selected_players)

team_xi = pd.concat(
    team_xi_results,
    ignore_index=True
)

print("Selected players:", len(team_xi))
print("Unique player-team selections:",
      team_xi[["team_name", "player_name"]].drop_duplicates().shape[0])

print("\nPlayers selected per team:")
display(team_xi.groupby("team_name").size())
```

```python
team_rating_results = []

for team in team_xi["team_name"].unique():

    xi = team_xi[team_xi["team_name"] == team]

    attack = xi[xi["position"].isin(["F", "M"])].nlargest(
        3, "Attack_Role_Score"
    )

    # Use the actual previously selected groups
    # to avoid selecting the same player twice.
    attack_players = set(attack["player_name"])

    midfield = xi[
        (xi["position"] == "M")
        & (~xi["player_name"].isin(attack_players))
    ]

    defence = xi[xi["position"] == "D"]
    goalkeeper = xi[xi["position"] == "G"]

    attack_rating = attack["Attack_Role_Score"].mean()
    midfield_rating = midfield["Midfield_Role_Score"].mean()
    defence_rating = defence["Overall_Rating"].mean()
    goalkeeper_rating = goalkeeper["Overall_Rating"].mean()

    overall = (
        0.30 * attack_rating
        + 0.25 * midfield_rating
        + 0.25 * defence_rating
        + 0.20 * goalkeeper_rating
    )

    team_rating_results.append({
        "team_name": team,
        "Attack_Rating": attack_rating,
        "Midfield_Rating": midfield_rating,
        "Defence_Rating": defence_rating,
        "Goalkeeping_Rating": goalkeeper_rating,
        "Overall_Team_Rating": overall
    })

team_ratings_final = pd.DataFrame(team_rating_results)

team_ratings_final = team_ratings_final.sort_values(
    "Overall_Team_Rating",
    ascending=False
).reset_index(drop=True)

team_ratings_final.index += 1

display(team_ratings_final.round(2))
```
