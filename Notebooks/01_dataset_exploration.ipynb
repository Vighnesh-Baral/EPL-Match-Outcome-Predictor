```python
# Define the features for our ML model

feature_columns = [
    "HomeTeam",
    "AwayTeam",
    "Home_Last5_Points",
    "Away_Last5_Points",
    "Home_Last5_GoalsScored",
    "Away_Last5_GoalsScored",
    "Home_Last5_GoalsConceded",
    "Away_Last5_GoalsConceded"
]

target_column = "FTR"

# Remove matches with unavailable historical features
model_df = match_features.dropna(
    subset=feature_columns
).copy()

print("Original matches:", len(match_features))
print("Usable matches:", len(model_df))
print("Removed matches:", len(match_features) - len(model_df))
```

```python
# Identify the latest two seasons

seasons = sorted(model_df["Season"].unique())

test_seasons = seasons[-2:]

# Chronological split
train_df = model_df[
    ~model_df["Season"].isin(test_seasons)
].copy()

test_df = model_df[
    model_df["Season"].isin(test_seasons)
].copy()

# Separate features and target
X_train = train_df[feature_columns]
y_train = train_df[target_column]

X_test = test_df[feature_columns]
y_test = test_df[target_column]

print("Training seasons:", seasons[:-2])
print("Testing seasons:", test_seasons)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Identify categorical and numerical features

categorical_features = [
    "HomeTeam",
    "AwayTeam"
]

numerical_features = [
    col for col in feature_columns
    if col not in categorical_features
]

# Encode team names
preprocessor = ColumnTransformer(
    transformers=[
        (
            "teams",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numbers",
            "passthrough",
            numerical_features
        )
    ]
)

# Create Random Forest
rf_model = RandomForestClassifier(
    n_estimators=500,
    max_features="sqrt",
    min_samples_leaf=3,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

# Complete ML pipeline
model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", rf_model)
])

print("Random Forest pipeline created successfully!")
```

```python
import pandas as pd
from collections import defaultdict

# Sort matches chronologically
elo_df = match_features.copy()

elo_df["Date"] = pd.to_datetime(elo_df["Date"])
elo_df = elo_df.sort_values(["Date"]).copy()

# Elo configuration
INITIAL_ELO = 1500
K_FACTOR = 20
HOME_ADVANTAGE = 60

# Store team ratings
ratings = defaultdict(lambda: INITIAL_ELO)

elo_rows = []

# Process matches date by date
for date, day_matches in elo_df.groupby("Date", sort=True):

    daily_updates = defaultdict(float)

    # Calculate pre-match ratings for every match on this date
    for idx, match in day_matches.iterrows():

        home = match["HomeTeam"]
        away = match["AwayTeam"]

        home_elo = ratings[home]
        away_elo = ratings[away]

        # Expected home and away performance
        expected_home = 1 / (
            1 + 10 ** ((away_elo - home_elo - HOME_ADVANTAGE) / 400)
        )

        expected_away = 1 - expected_home

        # Actual match result
        if match["FTR"] == "H":
            actual_home, actual_away = 1, 0

        elif match["FTR"] == "A":
            actual_home, actual_away = 0, 1

        else:
            actual_home, actual_away = 0.5, 0.5

        # Calculate rating changes
        home_change = K_FACTOR * (
            actual_home - expected_home
        )

        away_change = K_FACTOR * (
            actual_away - expected_away
        )

        daily_updates[home] += home_change
        daily_updates[away] += away_change

        elo_rows.append({
            "MatchIndex": idx,
            "Home_Elo": home_elo,
            "Away_Elo": away_elo,
            "Elo_Difference": (
                home_elo + HOME_ADVANTAGE - away_elo
            )
        })

    # Update ratings only after all matches on this date
    for team, change in daily_updates.items():
        ratings[team] += change

# Create Elo feature DataFrame
elo_features = pd.DataFrame(elo_rows).set_index("MatchIndex")

elo_features = elo_features.reindex(match_features.index)

# Add features to original dataset
match_features[
    ["Home_Elo", "Away_Elo", "Elo_Difference"]
] = elo_features[
    ["Home_Elo", "Away_Elo", "Elo_Difference"]
]

print("Elo features generated successfully!")

display(
    match_features[
        ["Date", "HomeTeam", "AwayTeam",
         "Home_Elo", "Away_Elo", "Elo_Difference"]
    ].head(10)
)
```

```python
# Updated feature selection with Elo ratings

feature_columns = [
    "HomeTeam",
    "AwayTeam",
    "Home_Last5_Points",
    "Away_Last5_Points",
    "Home_Last5_GoalsScored",
    "Away_Last5_GoalsScored",
    "Home_Last5_GoalsConceded",
    "Away_Last5_GoalsConceded",
    "Home_Elo",
    "Away_Elo",
    "Elo_Difference"
]

target_column = "FTR"

# Prepare updated dataset
model_df = match_features.dropna(
    subset=feature_columns + [target_column]
).copy()

print("Updated dataset shape:", model_df.shape)
print("Features:", len(feature_columns))
print("Missing values:", model_df[feature_columns].isna().sum().sum())
```

```python
# Identify the latest two seasons
seasons = sorted(model_df["Season"].unique())

test_seasons = seasons[-2:]

train_df = model_df[
    ~model_df["Season"].isin(test_seasons)
].copy()

test_df = model_df[
    model_df["Season"].isin(test_seasons)
].copy()

# Separate features and target
X_train = train_df[feature_columns]
y_train = train_df[target_column]

X_test = test_df[feature_columns]
y_test = test_df[target_column]

print("Training seasons:", train_df["Season"].unique())
print("Testing seasons:", test_seasons)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

categorical_features = ["HomeTeam", "AwayTeam"]

numeric_features = [
    col for col in feature_columns
    if col not in categorical_features
]

preprocessor = ColumnTransformer(
    transformers=[
        ("categorical",
         Pipeline([
             ("imputer", SimpleImputer(strategy="most_frequent")),
             ("encoder", OneHotEncoder(handle_unknown="ignore"))
         ]),
         categorical_features),

        ("numeric",
         SimpleImputer(strategy="median"),
         numeric_features)
    ]
)

rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=12,
    min_samples_split=10,
    min_samples_leaf=4,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", rf_model)
])

# Train
model.fit(X_train, y_train)

print("Random Forest with Elo training completed!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

y_pred = model.predict(X_test)

elo_accuracy = accuracy_score(y_test, y_pred)

print("Random Forest + Elo Accuracy:",
      round(elo_accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```
